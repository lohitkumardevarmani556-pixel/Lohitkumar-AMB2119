```python
student = {
  "name":"Namith",
  "ID no":"AMB2144" ,
  "branch":"Bsc argriculture",
  "marks" : 78
}
print("original dictionary:", student)
student_copy = student.copy()
print("copied dictionary:", student_copy)
removed_value = student_copy.pop("marks")
print("Removed_value:", removed_value)
print("after popitem():", student)
student.update({"year":2,"branch":"AIMl"})
print("after update():", student)
student.clear()
print("after update():",student)
del student
```
