```python
import numpy as np
#Nutrient replacement for 3 crops (n, p ,k  kg/ha)
#rows: crops (wheat, corn,rice)
#column: Nutrients (n, p , k)
nutrients_needs=np.array([
      [120,60,80],
      [150,75,100],
      [100,50,70]
  ])
#available fertilizer composition (percentage of n, p , k)
#rows: fertilizer type
#column: nutrient content
fertilizer_composition=np.array([
      [0.20,0.10,0.15],
      [0.15,0.20,0.10],
      [0.10,0.15,0.25]
  ])
print("nutrient requirement(kg/ha):")
print("nutrients_needs")
print("\nfertilizer composition(%):")
print("fertilizer_composition")

#matrix transpose example
print("\nTransposed nutrients needs(nutrients*crops):")
print("nutrients_needs:T")

#calculate fertilizer amount
total_nutrients_per_crop=np.sum(nutrients_needs, axis=1)
print(f"\ntotal nutrients needed per crop;{total_nutrients_per_crop}kg/ha")

#matrix multiplication : calculate total nutrient delivery
fertilizer_application=np.array([100,100,100])
nutrients_delivered=np.dot(fertilizer_application,fertilizer_composition)
print(f"\nNutrients from 100kg each fertilizer:{nutrients_delivered}")

#calculate determinant(for checking if matrix is invertible)
det=np.linalg.det(fertilizer_composition)
print(f"\nDeterminant of fertilizer matrix: {det:.6f}")

#identity matrix example
scaling=np.eye(3)*1.1
enhanced_needs=np.dot(nutrients_needs, scaling)
print("\nEnhanced nutrient needs(10% increase):")
print(enhanced_needs)
```

```python
import numpy as np
#rainfall data(mm) over 12 months
rainfall=np.array([45,52,68,120,180,220,195,160,110,75,58,40])
yields=np.array([2.1,3.2,5.6,2.3,5.6,4.3,9.2,5.6,2.6,4.6,6.5,8.6])
print("===RAINFALL STATISTICS===")
print(f"mean:{np.mean(rainfall):.2f}mm")
print("fmedian:{np.median(rainfall):.2f}mm")
print(f"std dev:{np.std(rainfall):.2f}mm")
print(f"variance:{np.var(rainfall):.2f}")
print(f"min:{np.min(rainfall)}mm(Month{np.argmin(rainfall)+1})")
print(f"max:{np.max(rainfall)}mm(Month{np.argmax(rainfall)+1})")
print(f"Range:{np.ptp(rainfall)}mm")



print("\n===YIELD STATISTICS===")
print(f"Mean:{np.mean(yields):.2f}tons/ha")
print(f"Median:{np.median(yields):.2f}tons/ha")
print(f"Std Dev:{np.std(yields):.2f}tons/ha")


print("\n===PERCENTILES===")
percentiles=[25,50,75,90]
rainfall_percentiles=np.percentile(rainfall,percentiles)
for p,val in zip(percentiles,rainfall_percentiles):
  print(f"{p}th percentiles rainfall:{val:.2f}mm")


print("\n===CORRELATION ANALYSIS===")
correlation_matrix=np.corrcoef(rainfall,yields)
correlation=correlation_matrix[0,1]
print(f"Rainfall-Yield Correlation:{correlation:.4f}")
print("interpretation: Strong positive correlation"if correlation>0.7 else "Moderate correlation" if correlation >0.4 else "Weak correlation")


print("\n===CUMULATIVE RAINFALL===")
cumulative=np.cumsum(rainfall)
print("Month\tRainfall\tCumulative")
for month,(rain,cum)in enumerate(zip(rainfall,cumulative),1):
  print(f"{month}\t{rain}\t\t{cum}")
```
